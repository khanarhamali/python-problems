```python
"""3. User will input (2numbers).Write a program to swap the numbers"""

L = []

def swap(list):
  if len(L) < 2:
    print("List must contain at least two numbers to swap.")
    return L

  L[0], L[1] = L[1], L[0]
  return L

a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

L.append(a)
L.append(b)

print(f"Original numbers: {L}")
swapped_numbers = swap(L)
print(f"Swapped numbers: {swapped_numbers}")
```

```python
"""3. User will input (2numbers).Write a program to swap the numbers"""

def swap(a,b):
  temp = a
  a = b
  b = temp
  return a, b

a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

print(f"Original numbers: {a}, {b}")
swapped_numbers = swap(a,b)
print(f"Swapped numbers: {swapped_numbers}")
```

```python
"""10. Write a program that will take user input of cost price and selling price and determines whether its a loss or a profit"""


cost_price = int(input("Enter Cost Price: "))
selling_price = int(input("Enter Selling Price: "))

def profit_loss(cost, sell):
  if cost < 0 and sell < 0:
    return "Not Possible"
  if cost > sell:
    return ("You got", f'{sell-cost}', "You Sold the Product in Loss")
  else:
    return ("You earned", f'{sell-cost}', "You Sold the Product in Profit")

profit_loss(cost_price, selling_price)
```

```python
"""16. Write a program that will determine weather when the value of temperature and humidity is provided by the user.
TEMPERATURE(C)      HUMIDITY(%)      WEATHER


      >= 30                             >=90                Hot and Humid

      >= 30                             < 90                 Hot

      <30                                >= 90               Cool and Humid

      <30                                 <90                 Cool"""

humi = int(input("Enter Humidity: "))
temp = int(input("Enter Temperature: "))

def weather(humidity, temp):
  if temp >= 30 and humidity >=90:
    print("Hot and Humid")
  if temp >=30 and humidity < 90:
    print("Hot")
  if temp < 30 and humidity >= 90:
    print("Humid")
  if temp < 30 and humidity < 90:
    print("Cool")

weather(humi, temp)
```
